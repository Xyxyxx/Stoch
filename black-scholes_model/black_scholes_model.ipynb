# The Black Scholes Model
We have two assets priced at times $B_t$ and $S_t$. $B_t$ here is the price of some zero risk asset, while $S_t$ is the price of some stock. 

$B_t$ being a **zero-risk asset** means that
$$
dB_t = r_t B_t dt.
$$
Here, $r_t$ is allowed to be some random process, but the main thing is that $B$ is not driven by any random term $dW_t$.

Now, we can write
$$
\frac{dB_t}{dt} = r_t B_t,
$$
which then gives us that
$$
B_t = B_0 e^{\int_0^t r_s \, ds}.
$$

We then model $S_t$ as following the process
$$
dS_t = \mu(t,S_t) S_t dt + \sigma(t, S_t) S_t dW_t.
$$
The (deterministic) function $\sigma$ is called **volatility** of $S$ and $\mu$ is called the **local mean rate of return.**


The most important special case of the above model is when $r, \mu,$ and $\sigma$ are given by just constants. This gives the famous **Black Scholes Model**:
$$
\begin{cases}
& dB_t = r B_t dt, \\
& dS_t = \mu S_t dt + \sigma S_t d W_t.
\end{cases}
$$
